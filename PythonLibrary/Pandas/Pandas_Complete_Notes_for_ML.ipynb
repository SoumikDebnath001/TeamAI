## 1. What is Pandas?

The two fundamental Pandas data structures are:

- **Series** — one-dimensional labeled data.
- **DataFrame** — two-dimensional labeled tabular data.

### Why Pandas is important for ML

A machine-learning workflow commonly needs to:

1. Load a dataset.
2. Inspect the data.
3. Clean missing/incorrect values.
4. Select useful features.
5. Transform columns.
6. Detect and handle outliers(data value that is significantly diff from other data).
7. Aggregate(combine multiple value to produce a summary) and analyze data.
8. Prepare data for a model.

Pandas is especially useful for steps **1–7**.

# 15. Missing Values

Missing data is extremely common in real-world datasets.

Important functions:

- `isna()` / `isnull()` → identify missing values
- `notna()` → identify non-missing values
- `dropna()` → remove missing data
- `fillna()` → replace missing data

### Important ML rule

Do **not blindly fill missing values**. The correct strategy depends on the feature and model.

Also, when building ML models, calculate imputation statistics from the **training data only** to avoid data leakage.


````python
missing = pd.DataFrame({
    "Age": [20, np.nan, 22, 21],
    "Score": [80, 90, np.nan, 70]
})

print(missing.isna())
print("\nMissing count:")
print(missing.isna().sum())

filled = missing.copy()
filled["Age"] = filled["Age"].fillna(filled["Age"].median())
filled["Score"] = filled["Score"].fillna(filled["Score"].mean())
filled
````

1. Split data into groups.
2. Apply an aggregation/transformation.
3. Combine the results.

````python
employees = pd.DataFrame({
    "Department": ["IT", "IT", "HR", "HR", "Sales"],
    "Salary": [50000, 70000, 45000, 55000, 60000],
    "Experience": [1, 3, 2, 4, 3]
})

employees.groupby("Department")["Salary"].mean()
````

# 38. Handling Outliers

Pandas can help identify outliers, but the statistical decision belongs to the analysis/modeling process.

A common IQR method:

- Q1 = 25th percentile
- Q3 = 75th percentile
- IQR = Q3 − Q1
- Lower bound = Q1 − 1.5 × IQR
- Upper bound = Q3 + 1.5 × IQR

Do not automatically delete every statistical outlier; some are legitimate observations.


````python
x = pd.Series([10, 11, 12, 13, 14, 100])

q1 = x.quantile(0.25)
q3 = x.quantile(0.75)
iqr = q3 - q1

lower = q1 - 1.5 * iqr
upper = q3 + 1.5 * iqr

outliers = x[(x < lower) | (x > upper)]

print("Lower:", lower)
print("Upper:", upper)
print("Outliers:", outliers.tolist())
````

Split the dataset into train/test **before learning transformations from the data** whenever the transformation can learn statistics from observations.

- `filter()` → keep/remove groups
- `apply()` → custom group operation

### Rule of thumb

Use `agg()` for summaries, `transform()` for features aligned to original rows, and `filter()` when deciding which groups remain.


````python
grouped = employees.groupby("Department")

print("Aggregate:")
print(grouped["Salary"].agg(["mean", "max"]))

print("\nTransform:")
print(grouped["Salary"].transform("mean"))

print("\nFilter groups with mean salary > 50000:")
print(grouped.filter(lambda g: g["Salary"].mean() > 50000))
````

# 60. `itertuples()` vs `iterrows()`

If iteration is genuinely necessary, `itertuples()` is usually preferable to `iterrows()` for performance and type preservation.

Still, a vectorized solution is normally better.


````python
small = pd.DataFrame({"A": [1, 2], "B": [3, 4]})

for row in small.itertuples(index=False):
    print(row.A, row.B)
````

# 72. Mini Practice Dataset

Use this dataset to practice the concepts above.

### Tasks

1. Inspect the dataset.
2. Find missing values.
3. Remove duplicate rows.
4. Convert `date` to datetime.
5. Calculate average salary by department.
6. Find employees with salary > 50,000.
7. Create an experience category.
8. Sort by salary.
9. Create a department-level average salary feature.
10. Export the cleaned dataset.


````python
practice = pd.DataFrame({
    "employee_id": [1, 2, 3, 4, 4, 5],
    "name": ["A", "B", "C", "D", "D", "E"],
    "department": ["IT", "HR", "IT", "Sales", "Sales", "HR"],
    "salary": [50000, 45000, np.nan, 65000, 65000, 55000],
    "experience": [1, 2, 3, 5, 5, 4],
    "date": ["2026-01-01", "2026-01-05", "2026-01-10",
             "2026-01-15", "2026-01-15", "2026-01-20"]
})

practice
````

````python
# Suggested solution

clean = (
    practice
    .drop_duplicates()
    .copy()
)

clean["date"] = pd.to_datetime(clean["date"])
clean["salary"] = clean["salary"].fillna(clean["salary"].median())

clean["experience_category"] = pd.cut(
    clean["experience"],
    bins=[-1, 2, 5, np.inf],
    labels=["Junior", "Mid", "Senior"]
)

clean["department_avg_salary"] = (
    clean.groupby("department")["salary"].transform("mean")
)

print(clean[clean["salary"] > 50000])
print("\nDepartment summary:")
print(clean.groupby("department")["salary"].mean())

clean = clean.sort_values("salary", ascending=False)
clean
````
